```python

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = {
    "order_id": [5001, 5002, 5003, 5004, 5005, 5006, 5007, 5008, 5009, 5010],
    "city": [
        "Lahore", "Karachi", "Islamabad", "Lahore", "Rawalpindi",
        "Karachi", "Islamabad", "Lahore", "Rawalpindi", "Karachi"
    ],
    "category": [
        "Electronics", "Clothing", "Electronics", "Books", "Clothing",
        "Books", "Clothing", "Electronics", "Books", "Electronics"
    ],
    "price": [150, 40, 2500, 15, 55, 20, 45, 180, 18, 160],
    "quantity": [2, 3, 1, 5, 2, 4, 1, 1, 6, 2],
    "rating": [4.5, 4.0, 3.5, 4.8, None, 4.2, 3.9, 4.1, 4.6, 3.8]
}

df = pd.DataFrame(data)
df
```

## Q19
### Answer :

```python
q1 = df["price"].quantile(0.25)
q3 = df["price"].quantile(0.75)
iqr = q3 - q1

lower_limit = q1 - 1.5 * iqr
upper_limit = q3 + 1.5 * iqr

print("Lower limit:", lower_limit)
print("Upper limit:", upper_limit)

outliers = df[
    (df["price"] < lower_limit) |
    (df["price"] > upper_limit)
]

print(outliers)
```

## Q20
### Answer :

```python
df_clean = df.dropna(subset=["rating"])

print("New shape:", df_clean.shape)

average_rating = (
    df_clean.groupby("city")["rating"]
    .mean()
    .sort_values(ascending=False)
)

print(average_rating)
```
