# Function For Calculating grades

```python
def calculate_marks():
    marks1 = float(input("Enter marks for Subject 1: "))
    marks2 = float(input("Enter marks for Subject 2: "))
    marks3 = float(input("Enter marks for Subject 3: "))
    marks4 = float(input("Enter marks for Subject 4: "))
    marks5 = float(input("Enter marks for Subject 5: "))

    total = marks1 + marks2 + marks3 + marks4 + marks5
    average = total / 5
    percentage = (total / 500) * 100

    print("Total Marks:", total)
    print("Average:", average)
    print("Percentage:", percentage, "%")
```

# Simple Calculator Using Functions

```python
def calculator():
    num1 = float(input("Enter first number: "))
    operator = input("Enter operator (+, -, *, /): ")
    num2 = float(input("Enter second number: "))

    if operator == "+":
        result = num1 + num2
    elif operator == "-":
        result = num1 - num2
    elif operator == "*":
        result = num1 * num2
    elif operator == "/":
        if num2 != 0:
            result = num1 / num2
        else:
            print("Cannot divide by zero")
            return
    else:
        print("Invalid operator")
        return

    print("Result:", result)
```

```python
calculator()
```

# Students Records Managemnt 

```python
###### students = {}


def add_student():
    name = input("Enter student name: ")
    marks = float(input("Enter marks: "))
    students[name] = marks
    print("Student added successfully!")


def update_marks():
    name = input("Enter student name: ")

    if name in students:
        marks = float(input("Enter new marks: "))
        students[name] = marks
        print("Marks updated successfully!")
    else:
        print("Student not found.")


def delete_student():
    name = input("Enter student name: ")

    if name in students:
        del students[name]
        print("Student deleted successfully!")
    else:
        print("Student not found.")


def search_student():
    name = input("Enter student name: ")

    if name in students:
        print("Name:", name)
        print("Marks:", students[name])
    else:
        print("Student not found.")


def display_students():
    if len(students) == 0:
        print("No student records found.")
    else:
        print("\nStudent Records:")
        for name, marks in students.items():
            print("Name:", name, "| Marks:", marks)


while True:
    print("\n--- Student Management System ---")
    print("1. Add Student")
    print("2. Update Marks")
    print("3. Delete Student")
    print("4. Search Student")
    print("5. Display All Students")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_student()
    elif choice == "2":
        update_marks()
    elif choice == "3":
        delete_student()
    elif choice == "4":
        search_student()
    elif choice == "5":
        display_students()
    elif choice == "6":
        print("Program ended.")
        break
    else:
        print("Invalid choice!")
```

# List Statistics

```python
def analyze_numbers():
    numbers = list(map(float, input("Enter numbers separated by spaces: ").split()))

    if len(numbers) < 2:
        print("Please enter at least 2 numbers.")
        return

    largest = max(numbers)
    smallest = min(numbers)
    total = sum(numbers)
    average = total / len(numbers)

    unique_numbers = list(set(numbers))
    unique_numbers.sort(reverse=True)

    if len(unique_numbers) >= 2:
        second_largest = unique_numbers[1]
    else:
        second_largest = "Not available"

    print("Largest number:", largest)
    print("Smallest number:", smallest)
    print("Sum:", total)
    print("Average:", average)
    print("Second largest number:", second_largest)


```

```python
analyze_numbers()
```

# Word Frequency COunter

```python
def count_words():
    sentence = input("Enter a sentence: ")

    words = sentence.split()
    word_count = {}

    for word in words:
        if word in word_count:
            word_count[word] += 1
        else:
            word_count[word] = 1

    return word_count
```

```python
result = count_words()

print("Word Count:")
print(result)

```

# Password Validator

```python
def check_password():
    password = input("Enter your password: ")

    has_upper = False
    has_lower = False
    has_digit = False
    has_special = False

    for char in password:
        if char.isupper():
            has_upper = True
        elif char.islower():
            has_lower = True
        elif char.isdigit():
            has_digit = True
        else:
            has_special = True

    if len(password) >= 8 and has_upper and has_lower and has_digit and has_special:
        print("Password is Strong and Valid")
    else:
        print("Password is Invalid")

        if len(password) < 8:
            print("- Password must have at least 8 characters")
        if not has_upper:
            print("- Password must contain an uppercase letter")
        if not has_lower:
            print("- Password must contain a lowercase letter")
        if not has_digit:
            print("- Password must contain at least one digit")
        if not has_special:
            print("- Password must contain at least one special character")
```

```python
check_password()
```

# Shoping Cart System
create a shopping cart using dictionary where keys are item names a nd valus are quantitiese ill using predefined item pricesAllow the user to  add item Remove items Update quantitise DIsplay the cart calculate total bu

```python
# Predefined item prices
prices = {
    "apple": 100,
    "banana": 50,
    "milk": 200,
    "bread": 150,
    "eggs": 300
}

# Shopping cart: item name -> quantity
cart = {}


def add_item():
    item = input("Enter item name: ").lower()

    if item in prices:
        quantity = int(input("Enter quantity: "))

        if item in cart:
            cart[item] += quantity
        else:
            cart[item] = quantity

        print("Item added successfully!")
    else:
        print("Item is not available.")


def remove_item():
    item = input("Enter item to remove: ").lower()

    if item in cart:
        del cart[item]
        print("Item removed successfully!")
    else:
        print("Item not found in cart.")


def update_quantity():
    item = input("Enter item name: ").lower()

    if item in cart:
        quantity = int(input("Enter new quantity: "))

        if quantity > 0:
            cart[item] = quantity
            print("Quantity updated successfully!")
        else:
            del cart[item]
            print("Item removed from cart.")
    else:
        print("Item not found in cart.")


def display_cart():
    if not cart:
        print("Your cart is empty.")
        return

    print("\n--- Shopping Cart ---")

    for item, quantity in cart.items():
        price = prices[item]
        subtotal = price * quantity

        print(item, "x", quantity, "=", subtotal)


def calculate_total():
    total = 0

    for item, quantity in cart.items():
        total += prices[item] * quantity

    print("Total Bill:", total)


while True:
    print("\n--- Shopping Cart Menu ---")
    print("1. Add Item")
    print("2. Remove Item")
    print("3. Update Quantity")
    print("4. Display Cart")
    print("5. Calculate Total")
    print("6. Exit")

    choice = input("Enter your choice: ")

    if choice == "1":
        add_item()
    elif choice == "2":
        remove_item()
    elif choice == "3":
        update_quantity()
    elif choice == "4":
        display_cart()
    elif choice == "5":
        calculate_total()
    elif choice == "6":
        print("Thank you for shopping!")
        break
    else:
        print("Invalid choice!")
```

# Exception handling challenge
write a programe that safely accepts two numbers from the user and perform  division ,handle possible exception such as invalid input division by zero use try except-finally

```python
def divide_numbers():
    try:
        num1 = float(input("Enter first number: "))
        num2 = float(input("Enter second number: "))

        result = num1 / num2

        print("Result:", result)

    except ValueError:
        print("Invalid input! Please enter numbers only.")

    except ZeroDivisionError:
        print("Error! Cannot divide by zero.")

    finally:
        print("Program finished.")
```

```python
divide_numbers()
```

# Libray Managment Mini Project 
Build a simple libray mangment system using lisdts ,dictionarise,functions,loops and exception handling,the system should allow users to Add books search books Borrow books Return books Display all available books

books = [
    {"title": "Python Basics", "author": "Ali", "available": True},
    {"title": "Data Science", "author": "Ahmed", "available": True},
    {"title": "Web Development", "author": "Sara", "available": False}
]


def add_book():
    title = input("Enter book title: ")
    author = input("Enter author name: ")

    book = {
        "title": title,
        "author": author,
        "available": True
    }

    books.append(book)
    print("Book added successfully!")


def search_book():
    title = input("Enter book title to search: ").lower()

    found = False

    for book in books:
        if book["title"].lower() == title:
            print("Book found!")
            print("Title:", book["title"])
            print("Author:", book["author"])

            if book["available"]:
                print("Status: Available")
            else:
                print("Status: Borrowed")

            found = True
            break

    if not found:
        print("Book not found.")


def borrow_book():
    title = input("Enter book title to borrow: ").lower()

    for book in books:
        if book["title"].lower() == title:

            if book["available"]:
                book["available"] = False
                print("Book borrowed successfully!")
            else:
                print("Sorry, this book is already borrowed.")

            return

    print("Book not found.")


def return_book():
    title = input("Enter book title to return: ").lower()

    for book in books:
        if book["title"].lower() == title:

            if not book["available"]:
                book["available"] = True
                print("Book returned successfully!")
            else:
                print("This book was not borrowed.")

            return

    print("Book not found.")


def display_available_books():
    print("\n--- Available Books ---")

    found = False

    for book in books:
        if book["available"]:
            print("Title:", book["title"])
            print("Author:", book["author"])
            print("----------------------")
            found = True

    if not found:
        print("No books are currently available.")





```python
while True:

    print("\n===== Library Management System =====")
    print("1. Add Book")
    print("2. Search Book")
    print("3. Borrow Book")
    print("4. Return Book")
    print("5. Display Available Books")
    print("6. Exit")

    try:
        choice = int(input("Enter your choice: "))

        if choice == 1:
            add_book()

        elif choice == 2:
            search_book()

        elif choice == 3:
            borrow_book()

        elif choice == 4:
            return_book()

        elif choice == 5:
            display_available_books()

        elif choice == 6:
            print("Thank you for using the library system!")
            break

        else:
            print("Please enter a number from 1 to 6.")

    except ValueError:
        print("Invalid input! Please enter a number.")
```
