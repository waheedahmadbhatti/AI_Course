## Funtion with return type

```python
def states(math,phy,com,eng):
    obt=math+phy+com+eng
    avg=obt/4
    percent=obt*100/350
    print(f"Obtained Marks ={obt}")
    print(f"Verage Marks ={avg}")
    print(f"Percentage Marks ={percent}")
    return obt,avg,percent
```

```python
obt,avg,percent=states(67,67,87,90)
```
