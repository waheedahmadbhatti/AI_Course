# OOP IN PYTHON

```python
class car:
    def __init__(self,model,color):
        self.model=model
        self.color=color

    def start(self):
        print(f"{self.model} Started")

    def stop(self):
        print(f"{self.model} Stop")

    def info(self):
        print(f"Model  {self.model} \n {self.color}")

    def change_color(self,color):
        self.color=color
        print("Color CHanged Successfully")
```

```python
car1=car('Honda','Red')
```

```python
car1.start()
```

```python
car1.stop()
```

```python
car1.info()
```
