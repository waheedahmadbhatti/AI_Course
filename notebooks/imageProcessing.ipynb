# Step 1: Import libraries and load the image

```python
import numpy as np
import matplotlib.pyplot as plt
from PIL import Image

# Load image
img = Image.open("C:/Users/Waheed/Downloads/secret.png")

# Convert image into NumPy array
img_array = np.array(img)

print("Image Shape:", img_array.shape)
print("Minimum Pixel:", img_array.min())
print("Maximum Pixel:", img_array.max())
print("Unique Pixel Values:", np.unique(img_array))

plt.imshow(img_array)
plt.axis("off")
plt.show()
```

# Step 2: Enhance image contrast

```python
# Convert to grayscale
gray = np.mean(img_array, axis=2)

# Enhance contrast
enhanced = (255 - gray) * 100

plt.figure(figsize=(10, 8))
plt.imshow(enhanced, cmap="gray")
plt.title("Enhanced Image")
plt.axis("off")
plt.show()
```

# Step 3: Separate RGB channels

```python
red = img_array[:, :, 0]
green = img_array[:, :, 1]
blue = img_array[:, :, 2]

fig, axes = plt.subplots(1, 3, figsize=(15, 5))

axes[0].imshow(red, cmap="gray")
axes[0].set_title("Red Channel")

axes[1].imshow(green, cmap="gray")
axes[1].set_title("Green Channel")

axes[2].imshow(blue, cmap="gray")
axes[2].set_title("Blue Channel")

for ax in axes:
    ax.axis("off")

plt.show()
```

# Step 4: Check bit planes (important)

```python
plt.figure(figsize=(12, 8))

for i in range(8):
    bit_plane = (blue >> i) & 1

    plt.subplot(2, 4, i + 1)
    plt.imshow(bit_plane, cmap="gray")
    plt.title(f"Blue Bit Plane {i}")
    plt.axis("off")

plt.tight_layout()
plt.show()
```

# Step 5: Find unusual pixel values

```python
unique, counts = np.unique(
    img_array.reshape(-1, 3),
    axis=0,
    return_counts=True
)

for pixel, count in zip(unique, counts):
    print("Pixel:", pixel, "Count:", count)
```

# Step 6: Extract pixels that are not pure white

```python
mask = np.any(img_array != 255, axis=2)

hidden_pixels = img_array[mask]

print("Unusual pixel count:", len(hidden_pixels))
print("Unusual pixel values:")
print(np.unique(hidden_pixels, axis=0))
```
